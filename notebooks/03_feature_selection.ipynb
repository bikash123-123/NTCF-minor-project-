# Feature Selection

This notebook documents the feature selection methodology used in the Network Threat Cognition Framework (NTCF). The objective is to identify the most relevant network traffic features for machine learning models using a reusable feature selection module.

## 1. Introduction

In this notebook we will:

- Load the preprocessed dataset.
- Compute feature importance using a Random Forest classifier.
- Rank features according to importance.
- Select the top 20 features.
- Save the selected feature names to `selected_features.json`.
- Verify the saved feature list.

```python
import os
import sys

# Go from notebooks/ to the project root
project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)

print("Project root:", project_root)
```

```python
import pandas as pd

from ml.preprocessing.preprocessing_pipeline import preprocess_dataset

from ml.feature_selection.feature_selection import (
    compute_feature_importance,
    select_top_features,
    save_selected_features,
    load_selected_features,
)
```

## 2. Load the Preprocessed Dataset

```python
X_train, X_test, y_train, y_test, encoder, scaler = preprocess_dataset()

print('Training feature shape :', X_train.shape)
print('Testing feature shape  :', X_test.shape)
print('Training labels shape  :', y_train.shape)
print('Testing labels shape   :', y_test.shape)
```

## 3. Compute Feature Importance

Compute feature importance using a Random Forest classifier trained on the preprocessed training dataset.

```python
importance = compute_feature_importance(
    X_train,
    y_train,
)

importance.head(20)
```

## 4. Display Feature Importance

```python
importance
```

## 5. Select Top 20 Features

```python
selected_features = select_top_features(
    importance,
    top_n=20
)

print('Selected Features:')
for feature in selected_features:
    print(feature)

print(f'\nTotal Selected Features: {len(selected_features)}')
```

## 6. Save Selected Features

```python
save_selected_features(selected_features)

print('Selected features saved successfully to:')
print('ml/feature_selection/selected_features.json')
```

## 7. Verify Saved Features

```python
loaded_features = load_selected_features()

print('Loaded Features:')
for feature in loaded_features:
    print(feature)
```

## 8. Conclusion

Feature importance was analyzed using a Random Forest classifier. The twenty most important features were selected and saved for reuse in the machine learning pipeline. The selected features remain compatible with the preprocessing pipeline and can be loaded during model training or future inference.
