# Support Vector Machine (SVM) Training

## Objective

This notebook demonstrates training a Support Vector Machine (SVM) classifier for the Network Threat Cognition Framework (NTCF).

The notebook performs the following tasks:

- Load the preprocessed dataset
- Train an SVM classifier
- Generate predictions
- Evaluate model performance
- Save the trained model
- Load the saved model
- Verify the loaded model

```python
import sys
from pathlib import Path

# Get the project root directory
project_root = Path.cwd().parent

# Add project root to Python path
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

print(project_root)
```

```python
import pandas as pd

from sklearn.metrics import confusion_matrix

from ml.preprocessing.preprocessing_pipeline import preprocess_dataset

from ml.models.svm_model import (
    create_model,
    train_model,
    predict,
    evaluate_model,
    save_model,
    load_model,
)
```

## Load Preprocessed Dataset

Load the cleaned, encoded, and scaled dataset generated by the preprocessing pipeline.

```python
X_train, X_test, y_train, y_test, encoder, scaler = preprocess_dataset()

print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print("Training Labels :", y_train.shape)
print("Testing Labels  :", y_test.shape)
```

## Create SVM Model

```python
model = create_model()

print(model)
```

## Train the Model

```python
model = train_model(
    model,
    X_train,
    y_train,
)

print("Model trained successfully.")
```

## Generate Predictions

```python
predictions = predict(
    model,
    X_test,
)

predictions[:20]
```

## Evaluate the Model

```python
results = evaluate_model(
    y_test,
    predictions,
)

print(f"Accuracy: {results['accuracy']:.4f}")
```

## Classification Report

```python
report = pd.DataFrame(
    results["classification_report"]
).transpose()

report
```

## Confusion Matrix

```python
cm = confusion_matrix(
    y_test,
    predictions,
)

print(cm)
```

## Save the Trained Model

```python
save_model(model)

print("SVM model saved successfully.")
```

## Load the Saved Model

```python
loaded_model = load_model()

print(type(loaded_model))
```

## Verify the Loaded Model

```python
loaded_predictions = predict(
    loaded_model,
    X_test,
)

loaded_predictions[:20]
```

```python
print(
    "Predictions identical:",
    (predictions == loaded_predictions).all()
)
```
